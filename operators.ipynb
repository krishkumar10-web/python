```python
# 1. Standard Assignment (=)
x = 10
print(x)

# 2. Add and Assign (+=)
x += 5  # Same as: x = x + 5
print(x)

# 3. Subtract and Assign (-=)
x -= 3  # Same as: x = x - 3
print(x)

# 4. Multiply and Assign (*=)
x *= 2  # Same as: x = x * 2
print(x)

# 5. Divide and Assign (/=)
x /= 4  # Same as: x = x / 4
print(x)

# 6. Floor Divide and Assign (//=)
x = 7
x //= 2  # Same as: x = x // 2
print(x)

# 7. Modulus and Assign (%=)
x = 10
x %= 3  # Same as: x = x % 3
print(x)

# 8. Exponentiate and Assign (**=)
x = 2
x **= 3  # Same as: x = 2 * 2 * 2
print(x)
```

```python
# 3. Comparison (Relational) Operators
# 1. Equal to (==)
# Checks if two values are exactly the same.
print("Is 5 equal to 5?", 5 == 5)
print("Is 5 equal to 3?", 5 == 3)

# 2. Not Equal to (!=)
# Checks if two values are different.
print("Is 5 different from 3?", 5 != 3)
print("Is 5 different from 5?", 5 != 5)

# 3. Greater than (>)
# Checks if the left number is bigger.
print("Is 10 greater than 5?", 10 > 5)

# 4. Less than (<)
# Checks if the left number is smaller.
print("Is 3 less than 8?", 3 < 8)

# 5. Greater than or equal to (>=)
# Checks if the left number is bigger OR exactly the same.
print("Is 5 greater than or equal to 5?", 5 >= 5)
print("Is 6 greater than or equal to 5?", 6 >= 5)

# 6. Less than or equal to (<=)
# Checks if the left number is smaller OR exactly the same.
print("Is 4 less than or equal to 4?", 4 <= 4)
```

```python

# IDENTITY OPERATORS (is, is not)
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a  

# 1. 'is' operator
print(list_a is list_b)
print(list_a is list_c)

# 2. 'is not' operator
print(list_a is not list_b)


```

```python

# MEMBERSHIP OPERATORS (in, not in)



my_fruits = ["apple", "banana", "cherry"]
my_text = "Hello world"

# 1. 'in' operator
print("banana" in my_fruits)
print("apple" in my_fruits)
print("world" in my_text)

# 2. 'not in' operator
print("grape" not in my_fruits)



```

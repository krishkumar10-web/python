# Python String Methods

Easy notes with one-line explanations and simple examples.

```python
# isidentifier() checks if a string can be used as a valid Python variable name.
text = "student_name"
print(text.isidentifier())
```

```python
# isprintable() checks if all characters in a string can be printed normally.
text = "Hello World!"
print(text.isprintable())
```

```python
# isascii() checks if all characters in a string are ASCII characters.
text = "Hello123"
print(text.isascii())

```

```python
# casefold() converts a string to a lowercase form for case-insensitive comparison.
text = "PYTHON"
print(text.casefold())
```

```python
# expandtabs() replaces tab characters (\t) with spaces.
text = "Name:\tKrish"
print(text.expandtabs(4))
```

```python
# join() joins multiple strings together using a separator.
words = ["I", "love", "Python"]
print(" ".join(words))
```

```python
# partition() divides a string into three parts at the first matching separator.
text = "name=Krish"
print(text.partition("="))
```

```python
# rpartition() divides a string into three parts at the last matching separator.
text = "a-b-c"
print(text.rpartition("-"))
```

```python
# splitlines() splits a string into a list using line breaks.
text = "Hello\nPython\nWorld"
print(text.splitlines())
```

```python
# rindex() returns the position of the last occurrence of a substring; it gives an error if not found.
text = "banana"
print(text.rindex("n"))
```

```python
# rfind() returns the position of the last occurrence of a substring; it returns -1 if not found.
text = "banana"
print(text.rfind("a"))
```
